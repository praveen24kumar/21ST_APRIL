```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Display DataFrame info
print(cinePOS_theaters.info())

# Initialize encoder
enc = OneHotEncoder(handle_unknown='ignore', sparse=False)

# Get list of categorical columns (excluding first one if intentional)
categorical_columns = cinePOS_theaters.select_dtypes(include=['object']).columns.tolist()
print(categorical_columns[1:])  # Adjust if you want to include all

# Fit and transform the selected categorical columns
one_hot_encoded = enc.fit_transform(cinePOS_theaters[categorical_columns[1:]])

# Create DataFrame from encoded array
df = pd.DataFrame(one_hot_encoded, columns=enc.get_feature_names_out(categorical_columns[1:]))

# Optional: Concatenate with original DataFrame and drop original categorical columns
df_cinePOS_theaters_encoded = pd.concat([cinePOS_theaters, df], axis=1).drop(categorical_columns[1:], axis=1)

print("cinePOS_theaters: ", cinePOS_theaters.shape)
print("df_cinePOS_theaters_encoded: ", df_cinePOS_theaters_encoded.shape)
```
