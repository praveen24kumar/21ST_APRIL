```python
print(booknow_visits.columns)
booknow_visits['show_date'] = pd.to_datetime(booknow_visits['show_date'])

booknow_visits['day'] = booknow_visits['show_date'].dt.day
booknow_visits['month'] = booknow_visits['show_date'].dt.month
booknow_visits['year'] = booknow_visits['show_date'].dt.year
booknow_visits['dayofyear'] = booknow_visits['show_date'].dt.dayofyear

booknow_visits = booknow_visits.drop(columns=['show_date'])

print(booknow_visits.head)
```

```python
from sklearn.model_selection import train_test_split

y = booknow_visits['audience_count']
X = booknow_visits.drop(columns=['audience_count'])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)
print(X_train.columns)
print(X_train.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

numerical_cols = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = X_train.select_dtypes(include=['object', 'category']).columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        ('num', MinMaxScaler(), numerical_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ])

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

```python
from sklearn.svm import LinearSVR
from sklearn.metrics import r2_score

# Initialize LinearSVR
model = LinearSVR(random_state=42)

# Train the model
model.fit(X_train_processed, y_train)

# Predict on test set
y_pred = model.predict(X_test_processed)

# Calculate R2 score
r2 = r2_score(y_test, y_pred)

# Print rounded R2 score
print(round(r2, 3))
```

```python
from sklearn.decomposition import TruncatedSVD
from sklearn.svm import LinearSVR
from sklearn.metrics import mean_absolute_error

#Applying TruncatedSVD on the processed features
svd = TruncatedSVD(n_components=4, random_state=42)

X_train_svd = svd.fit_transform(X_train_processed)
X_test_svd = svd.transform(X_test_processed)

#Training LinearSVR
model = LinearSVR(random_state=42)
model.fit(X_train_svd, y_train)

#Prediction on training data
y_train_pred = model.predict(X_train_svd)
y_pred = model.predict(X_test_svd)

#Computing Mean Absolute Error (MAE)
mae_train = mean_absolute_error(y_train, y_train_pred)

#Printing rounded MAE
print(round(mae_train, 3))

r2 = r2_score(y_test, y_pred)

#Printing rounded R2 score
print(round(r2, 3))
```

```python
from sklearn.feature_selection import SelectKBest, mutual_info_regression
import numpy as np

selector = SelectKBest(score_func=mutual_info_regression, k=2)

selector.fit(X_train_processed, y_train)

scores = selector.scores_

max_score = np.nanmax(scores)   # handles potential NaN values safely

print("Maximum score:", round(max_score, 3))
```

```python
mean_score = np.nanmean(scores)

print("Mean of scores:", round(mean_score, 3))
```
